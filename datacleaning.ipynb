```python
import pandas as pd
import numpy as np
import csv

# 1. LOAD THE RAW FILE

input_file = "bank dataset.csv"

rows = []

with open(input_file, "r", encoding="utf-8") as file:

    for line in file:

        line = line.strip()

        # Remove outer quotation marks
        if line.startswith('"') and line.endswith('"'):
            line = line[1:-1]

        # Split using ;
        row = line.split(";")

        # Remove quotation marks and spaces
        row = [
            value.strip().strip('"')
            for value in row
        ]

        rows.append(row)


# 2. CREATE DATAFRAME

columns = rows[0]

df = pd.DataFrame(
    rows[1:],
    columns=columns
)


# 3. CLEAN COLUMN NAMES

df.columns = (
    df.columns
    .str.strip()
    .str.replace('"', '', regex=False)
)


print("COLUMN NAMES:")
print(df.columns.tolist())

print("\nDATASET SHAPE:")
print(df.shape)

print("\nFIRST 5 ROWS:")
print(df.head())
```

```python

# 4. CHECK MISSING VALUES


print("\n--- MISSING VALUES ---")

print(df.isnull().sum())


# 5. CHECK DUPLICATES

print("\n--- DUPLICATES ---")

duplicate_count = df.duplicated().sum()

print("Duplicate records:", duplicate_count)


# 6. CHECK DATA TYPES BEFORE CLEANING

print("\n--- DATA TYPES BEFORE CLEANING ---")

print(df.dtypes)


# 7. REMOVE EXTRA SPACES


for column in df.select_dtypes(include="object").columns:

    df[column] = df[column].str.strip()



# 8. STANDARDIZE TEXT VALUES


for column in df.select_dtypes(include="object").columns:

    df[column] = df[column].str.lower()


# 9. HANDLE UNKNOWN / INCONSISTENT VALUES

df.replace(
    ["unknown", "?", "na", "n/a", ""],
    np.nan,
    inplace=True
)


# 10. REMOVE DUPLICATES

duplicates_before = df.duplicated().sum()

df = df.drop_duplicates()

print("\nDuplicates removed:", duplicates_before)


# 11. CONVERT NUMERIC COLUMNS

numeric_columns = [
    "age",
    "balance",
    "day",
    "duration",
    "campaign",
    "pdays",
    "previous"
]

for column in numeric_columns:

    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )


# 12. HANDLE MISSING NUMERIC VALUES


for column in numeric_columns:

    df[column] = df[column].fillna(
        df[column].median()
    )



# 13. HANDLE MISSING CATEGORICAL VALUES


categorical_columns = [
    "job",
    "marital",
    "education",
    "default",
    "housing",
    "loan",
    "contact",
    "month",
    "poutcome",
    "y"
]

for column in categorical_columns:

    df[column] = df[column].fillna(
        df[column].mode()[0]
    )



# 14. VALIDATE NUMERIC VALUES


# Age cannot be negative
df.loc[df["age"] < 0, "age"] = np.nan
df["age"] = df["age"].fillna(df["age"].median())


# Duration cannot be negative
df.loc[df["duration"] < 0, "duration"] = np.nan
df["duration"] = df["duration"].fillna(
    df["duration"].median()
)


# Campaign cannot be negative
df.loc[df["campaign"] < 0, "campaign"] = np.nan
df["campaign"] = df["campaign"].fillna(
    df["campaign"].median()
)


# Previous contacts cannot be negative
df.loc[df["previous"] < 0, "previous"] = np.nan
df["previous"] = df["previous"].fillna(
    df["previous"].median()
)


# 15. FINAL DATA QUALITY CHECK


print("\n====================================")
print("FINAL DATA QUALITY CHECK")
print("====================================")

print("\nFinal Shape:")
print(df.shape)

print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nRemaining Duplicate Records:")
print(df.duplicated().sum())

print("\nFinal Data Types:")
print(df.dtypes)



# 16. SAVE CLEANED DATASET

output_file = "bank_marketing_cleaned.csv"

df.to_csv(
    output_file,
    index=False
)

print("\n====================================")
print("CLEANING COMPLETED SUCCESSFULLY")
print("====================================")

print("Saved as:", output_file)
```
